```python
import pandas as pd

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("student_performance.csv")
```

```python
print("Shape of the dataset: ", df.shape)
print("Number of Students: ", df.shape[0])
print("Number of Columns: ", df.shape[1])
```

```python
print("Columns in the dataset")
print(df.columns.tolist())
```

```python
print("Data Types")
df.dtypes
```

```python
## Statistical Summary
#=========================
df.describe()
```

```python
## Check for Missing values
#===========================
print("Missing value in each column")
print(df.isnull().sum())

print("Vertical missing values", df.isnull().sum().sum())
```

```python
## View a specific column
#========================
print("Student Name")
print(df['name'])
```

```python
## View Unique Departments
#=========================

print("\033[1mDepartments in the dataset:\033[0m")
print(df['department'].unique())
print("\n\033[1mNumber of Departments:\033[0m")
print((df['department'].value_counts()))
```

```python
## Filter Students by Department
#===============================

cs_students = df[df['department'] == 'Computer Science']
print(f"Number of Computer Science Students: {len(cs_students)}")

print("\nComputer Science students:")
print(cs_students[['name', 'math_score', 'science_score', 'english_score', 'programming_score']])
```

```python
cs_students_high = df[df['attendance_percentage'] > 90]
print(cs_students_high[['name', 'attendance_percentage']])
```

```python
## Find the top Performing Students
#=================================

df['total_score'] = df['math_score'] + df['science_score'] + df['english_score'] + df['programming_score']
toppers = df[df['total_score'] / 4 > 90]

print("Top Performing Students:")
print(toppers[['name', 'total_score']])
```

```python
## Calculate average score per Department
#========================================
dept_avg = df.groupby('department')['total_score'].mean().round(2)
print(dept_avg)
```

```python
## Printing Department average in Descending order
#===============================================
print(dept_avg.sort_values(ascending=False))
```
