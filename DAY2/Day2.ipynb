**DATABASE FUNDAMENTALS + SQLITE3 FOR STUDENTS DATABASE**

```python
# Import Libraries
#=========================================

import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("All libraries imported successfully")
print(f"Pandas version: {pd.__version__}")
print(f"SQLite version: {sqlite3.version}")
```

```python
# Load the Student Dataset
#==================================

df = pd.read_csv('student_performance.csv')
print(f"Dataset Loaded: {df.shape[0]} students, {df.shape[1]} columns")
print(f"Columns: {df.columns.tolist()}")
df.head(10)
```

```python
# Create SQLite database and load Student Data
#===================================================

conn = sqlite3.connect('college.db')

cursor = conn.cursor()
print('Database Created Successfully')

df.to_sql(
    'students',
    conn,
    if_exists='replace',
    index=False
)
cursor.execute("SELECT COUNT(*) FROM students;")
count = cursor.fetchone()[0]
print(f"Number of rows in the 'students' table: {count}")
```

```python
select = cursor.execute("SELECT name, math_score FROM students WHERE math_score > 75")
results = select.fetchall()
df_results = pd.DataFrame(results, columns=['Name', 'Math Score'])
display(df_results)
```

```python
query = """
    SELECT student_id, name, department, math_score, attendance_percentage
    FROM students
    LIMIT 10
"""
cursor.execute(query)

results = cursor.fetchall()

df_result1 = pd.DataFrame(results, columns=['Student ID', 'Name', 'Department', 'Math Score', 'Attendance Percentage'])
display(df_result1)
```

```python
query2 = """
    SELECT name, Department, math_score
    FROM students
    ORDER BY math_score DESC
    LIMIT 5
"""
cursor.execute(query2)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Name','Department', 'Math Score'])
print(df_results2)
```

```python
#================================================
# Students From CS Department (WHERE Clause)
#================================================

query3 = """
    SELECT name, math_score, science_score, programming_score, attendance_percentage
    FROM students
    WHERE department = 'Computer Science'
    LIMIT 10
"""
cursor.execute(query3)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Name', 'Math Score', 'Science Score', 'Programming Score', 'Attendance Percentage'])
print(df_results2)
```

```python
#================================================
# Multiple WHERE Conditions
#================================================

query4 = """
    SELECT name, department, attendance_percentage
    FROM students
    WHERE department <> 'Civil' and attendance_percentage > 90
    ORDER BY attendance_percentage DESC
    LIMIT 10
"""
cursor.execute(query4)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Name', 'Department', 'Attendance Percentage'])
print(df_results2)
```

```python
#================================================
# BETWEEN, IN KEYWORDS
#================================================

query5 = """
    SELECT name, department, attendance_percentage
    FROM students
    WHERE name BETWEEN 'A' AND 'F'
    ORDER BY name
"""
cursor.execute(query5)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Name', 'Department', 'Attendance Percentage'])
print(df_results2)
```

```python
#=======================================================
# GROUP BY Clause with conditions (Filter by department)
#=======================================================

query6 = """
    SELECT
    department,
    COUNT(*) AS student_count,
    ROUND(AVG(math_score), 2) AS avg_math_score,
    ROUND(AVG(science_score), 2) AS avg_science_score,
    ROUND(AVG(programming_score), 2) AS avg_programming_score,
    ROUND(AVG(attendance_percentage), 2) AS avg_attendance
    FROM students
    GROUP BY department
    ORDER BY avg_math_score DESC
"""
cursor.execute(query6)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Department', 'Student Count', 'Average Math Score', 'Average Science Score', 'Average Programming Score', 'Average Attendace'])
print(df_results2.to_string(index=False))
```

```python
#================================================
# Departments with AVG Math > 70 (HAVING)
#================================================

query7 = """
    SELECT
    department,
    ROUND(AVG(math_score), 2) AS avg_math_score,
    COUNT(*) AS student_count
    FROM students
    GROUP BY department
    HAVING avg_math_score > 70
    ORDER BY avg_math_score DESC
"""
cursor.execute(query7)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Department', 'Average Math Score', 'Students'])
print(df_results2)
```

```python
#================================================
# Gender-wise Performance Analysis
#================================================

query8 = """
    SELECT
    gender,
    COUNT(*) AS total_students,
    ROUND(AVG(math_score), 2) AS avg_math_score,
    ROUND(AVG(programming_score), 2) AS avg_programming_score,
    ROUND(AVG(attendance_percentage), 2) AS avg_attendance
    FROM students
    GROUP BY gender
"""
cursor.execute(query8)
results = cursor.fetchall()
df_results2 = pd.DataFrame(results, columns=['Gender', 'Total Students', 'Average Math Score', 'Average Programming Score', 'Average Attendance'])
print(df_results2.to_string(index=False))
```
