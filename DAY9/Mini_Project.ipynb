# **AI Agents + Text-to-SQL**

```python
#=====================================================
# Question: Who are the top 5 students in Programming?
#=====================================================

import pandas as pd

df = pd.read_csv('student_performance.csv')
df.head()
```

### Step 1: Install and Setup

### Step 2: Load Data into SQLite

```python
import sqlite3

conn = sqlite3.connect('students.db')

df.to_sql('performance', conn, if_exists='replace', index=False)

print('Database created and table "performance" populated.')
```

### Step 3: Initialize Groq Client


```python
from groq import Groq
import os

os.environ['GROQ_API_KEY'] = '************************************'

client = Groq(api_key=os.environ.get('GROQ_API_KEY'))
```

### Step 4: Text-to-SQL Logic

```python
def get_sql_query(user_question):
    prompt = f"""
    You are a SQL expert. Given the table 'performance' with columns:
    {', '.join(df.columns)}

    Generate only the SQL query to answer this question: {user_question}
    Do not include any explanation or markdown formatting, just the raw SQL.
    """

    chat_completion = client.chat.completions.create(
        messages=[{"role": "user", "content": prompt}],
        model="llama-3.1-8b-instant",
    )
    return chat_completion.choices[0].message.content.strip()

question = "Who are the top 5 students in Programming?"
sql = get_sql_query(question)
print(f"Generated SQL: {sql}")
```

### Step 5: Final Result


```python
result_df = pd.read_sql_query(sql, conn)
display(result_df)
```
